```python
import pandas as pd

# ===============================
# File paths
# ===============================
input_file  = "feature_selected_dataset.csv"
output_file = "final_features_GOOGL_2022_2025.csv"

# ===============================
# Load data
# ===============================
df = pd.read_csv(input_file)

# ===============================
# Identify key columns
# ===============================
date_col = next(c for c in df.columns if "date" in c.lower())
stock_col = next(c for c in df.columns if "ticker" in c.lower())

# Convert date column to datetime
df[date_col] = pd.to_datetime(df[date_col])

# ===============================
# 1. Drop 'close' and 'volume'
# ===============================
df = df.drop(columns=[c for c in df.columns if c.lower() in ["close", "volume"]])

# ===============================
# 2. Keep only GOOGL
# ===============================
df = df[df[stock_col] == "GOOGL"]

# ===============================
# 3. Filter date range
# ===============================
start_date = pd.Timestamp("2022-01-01")
end_date   = pd.Timestamp("2025-12-31")

df = df[(df[date_col] >= start_date) & (df[date_col] <= end_date)]

# ===============================
# Save to NEW file (original untouched)
# ===============================
df.to_csv(output_file, index=False)
print("Saved cleaned dataset to:", output_file)
```
